```python
# ==============================================================================
# Statistical and Environmental Analyses of Realized Niche Breadth in Atta cephalotes and Atta laevigata
# @date Aug 10th, 2026
# @author Diego García Castillo
# 
# This notebook contains the data processing, statistical analyses, and visualization workflows 
# used to characterize and compare the realized niche breadth of Atta cephalotes and Atta laevigata 
# across the study area in Santander, Colombia.
# ==============================================================================

#Loading basic libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from skbio.stats.distance import permanova, permdisp
from skbio import DistanceMatrix
from scipy.spatial.distance import pdist, squareform
from matplotlib.lines import Line2D
import matplotlib.colors as mcolors
import geopandas as gpd
import contextily as cx
from matplotlib_scalebar.scalebar import ScaleBar # Import the scalebar tool
from matplotlib.patches import Patch
import matplotlib as mpl
from pyproj import Transformer
import matplotlib.patches as patches
import matplotlib.patheffects as pe
import cdsapi
```

```python
# --- LOAD THE DATASETS ---

#Fine-scale environmental variables as a result of the sampling
DATA_FIELDWORK_CSV = "../data/Ants_Fieldwork_Standard_Data.csv"

#Coordinates of the localities
DATA_FIELDWORK_LOCATION_COORDINATES = "../data/FieldworkAnts2026_LocationCoordinates.csv"
# 1. load data
df = pd.read_csv(DATA_FIELDWORK_CSV)

# --- PALETTE DEFINITION ---
species_palette = {'L': '#D97A2B', 'C': '#6FAF4F'} 

#Array of variables that we used in the analyses
array_variables = ['Mound_Area_m2', 'Shape_Ratio', 'Soil_pH', 'Light_Percentage', 'In_Temp', 'Temp_Diff']
```

```python
# 2. Time conversion to decimal
# Convert the text 'hh:mm' to time objects and then to decimal numbers
df['Time_Value'] = pd.to_datetime(df['Time_Photo'], format='%H:%M').dt.hour + \
                   pd.to_datetime(df['Time_Photo'], format='%H:%M').dt.minute / 60

# 3. Calculate thermic efficiency
# Delta T
df['Temp_Diff'] = df['In_Temp'] - df['Out_Temp']


# 4. Axis Standardization (Morphometric Correction)
# We ensure 'Length' is the major axis (max) and 'Width' is the minor axis (min)
df['Length_adj'] = df[['Length', 'Width']].max(axis=1)
df['Width_adj'] = df[['Length', 'Width']].min(axis=1)

# 5. Feature Engineering
# Calculate the Area of the mound assuming an Elliptical shape
# Area = pi * (Length/2) * (Width/2)
df['Mound_Area'] = np.pi * (df['Length_adj'] / 2) * (df['Width_adj'] / 2)

df['Mound_Area_Log'] = np.log(df['Mound_Area'])
df['Mound_Area_m2'] = np.pi * ((df['Length_adj']/100) / 2) * ((df['Width_adj']/100) / 2)

# 6. Calculate the Shape Ratio (Elongation Index)
# A ratio of 1.0 means a perfect circle; > 1.0 indicates an elongated ellipse
df['Shape_Ratio'] = df['Length_adj'] / df['Width_adj']

# 7. Remove U species, because it will not be considered in the analysis
df_final = df[df['Species'] != 'U'].copy()
species_list = df_final['Species'].unique()
```

```python
# ==============================================================================
# Identify whether the data follow a normal distribution or not
# ==============================================================================

def identify_normality(data_sp1, data_sp2):
    """
    Validates normality for two groups using Shapiro-Wilk.
    Returns distribution type and p-values for both groups.
    """
    data_sp1 = data_sp1.dropna()
    data_sp2 = data_sp2.dropna()
    
    if len(data_sp1) < 3 or len(data_sp2) < 3:
        return "Insufficient Data", np.nan, np.nan

    # 1. Validate normality (Shapiro-Wilk)
    _, p_norm1 = stats.shapiro(data_sp1)
    _, p_norm2 = stats.shapiro(data_sp2)
    
    # 2. Decision logic
    if p_norm1 > 0.05 and p_norm2 > 0.05:
        distribution = "Normal"
    else:
        distribution = "Non-normal"
    
    return distribution, p_norm1, p_norm2

# Results container
test_results = []

for var in array_variables:
    group_a = df_final[df_final['Species'] == species_list[0]][var]
    group_b = df_final[df_final['Species'] == species_list[1]][var]

    # Calculate Mean and SD for each species
    mean_a, sd_a = group_a.mean(), group_a.std()
    mean_b, sd_b = group_b.mean(), group_b.std()
    
    # Run the updated function
    dist_type, p_val_sp1, p_val_sp2 = identify_normality(group_a, group_b)
    
    # Append results with p-values rounded for neatness
    test_results.append({
        'Ecological Variable': var.replace('_', ' '),
        f'Mean ({species_list[0]})': round(mean_a, 3),
        f'SD ({species_list[0]})': round(sd_a, 3),
        f'Mean ({species_list[1]})': round(mean_b, 3),
        f'SD ({species_list[1]})': round(sd_b, 3),
        f'p-norm ({species_list[0]})': round(p_val_sp1, 4),
        f'p-norm ({species_list[1]})': round(p_val_sp2, 4),
        'Assumed Distribution': dist_type
    })

# Create summary DataFrame
df_stats_summary = pd.DataFrame(test_results)

# Professional renaming of variables
variable_map = {
    'Mound Area': 'Mound Area (cm²)',
    'Mound Area m2': 'Mound Area (m²)',
    'Shape Ratio': 'Shape Ratio',
    'Soil pH': 'Soil pH',
    'Light Percentage': 'Canopy Openness (%)',
    'In Temp': 'Internal Temperature (°C)',
    'Temp Diff': 'Thermal Regulation Capacity (°C)'
}
df_stats_summary['Ecological Variable'] = df_stats_summary['Ecological Variable'].map(variable_map)

print("--- Table S3. Statistical Normality Results ---")
print(df_stats_summary)

# Save to Excel
df_stats_summary.to_excel("Table_S2_Statistical_Normality_Results_with_Pvalues.xlsx", index=False)
```

```python
# ==============================================================================
# Run PERMANOVA for the variables that are non-normally distributed
# ==============================================================================

# Use a fixed seed for reproducibility
fixed_seed = 42

# List of variables to test
variables_to_test = [
    'Mound_Area_m2',
    'Shape_Ratio',
    'Soil_pH',
    'Light_Percentage',
    'Temp_Diff',
    'In_Temp'
]

stats_results = []

for var in variables_to_test:

    # 1. Clean data
    sub_df = df_final[[var, 'Species', 'Location']].dropna().copy()
    sub_df = sub_df.reset_index(drop=True)

    # 2. Calculate descriptive statistics using original values
    species_summary = (
        sub_df.groupby('Species')[var]
        .agg(['mean', 'std', 'count'])
    )

    mean_cephalotes = species_summary.loc['C', 'mean']
    sd_cephalotes = species_summary.loc['C', 'std']

    mean_laevigata = species_summary.loc['L', 'mean']
    sd_laevigata = species_summary.loc['L', 'std']

    # 3. Standardize the variable
    var_data = sub_df[[var]].values

    scaled_data = (
        var_data - var_data.mean()
    ) / var_data.std(ddof=0)

    # 4. Create Euclidean distance matrix
    dist_data = pdist(scaled_data, metric='euclidean')

    dist_matrix = DistanceMatrix(
        squareform(dist_data),
        ids=sub_df.index.astype(str)
    )

    sub_df.index = sub_df.index.astype(str)

    # 5. PERMANOVA for Species
    res_perm_species = permanova(
        dist_matrix,
        sub_df,
        column='Species',
        permutations=999,
        seed=fixed_seed
    )

    # 6. PERMANOVA for Location
    res_perm_location = permanova(
        dist_matrix,
        sub_df,
        column='Location',
        permutations=999,
        seed=fixed_seed
    )

    # 7. PERMDISP for Species
    try:
        res_disp_species = permdisp(
            dist_matrix,
            sub_df,
            column='Species',
            permutations=999,
            seed=fixed_seed
        )

        p_disp_sp = res_disp_species['p-value']

    except ZeroDivisionError:
        p_disp_sp = np.nan

    # 8. Calculate R² manually
    n = len(sub_df)

    # Species
    F_species = res_perm_species['test statistic']
    g_species = sub_df['Species'].nunique()

    R2_species = (
        F_species * (g_species - 1)
    ) / (
        F_species * (g_species - 1) + (n - g_species)
    )

    # Location
    F_location = res_perm_location['test statistic']
    g_location = sub_df['Location'].nunique()

    R2_location = (
        F_location * (g_location - 1)
    ) / (
        F_location * (g_location - 1) + (n - g_location)
    )

    # 9. Save results
    stats_results.append({
        'Variable': var,

        'A_cephalotes_mean': round(mean_cephalotes, 3),
        'A_cephalotes_SD': round(sd_cephalotes, 3),

        'A_laevigata_mean': round(mean_laevigata, 3),
        'A_laevigata_SD': round(sd_laevigata, 3),

        'F_stat_Species': round(F_species, 3),
        'R2_Species': round(R2_species, 3),

        'PERMANOVA_Species_p': res_perm_species['p-value'],
        'PERMDISP_Species_p': p_disp_sp,

        'PERMANOVA_Location_p': res_perm_location['p-value'],
        'R2_Location': round(R2_location, 3)
    })

# Create results table
df_stats_integrated = pd.DataFrame(stats_results)
```

```python
# ==============================================================================
# CREATE A BOXPLOT WITH THE DISTRIBUTION OF THE DATA PER VARIABLE PER LOCALITY
# ==============================================================================

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.lines import Line2D
import matplotlib.colors as mcolors

# Mapping dictionary for locations
location_codes = {
    'Montefrio': 'MF', 'Nemizaque': 'NZ', 'Morario': 'MO', 'Villas': 'VI',
    'Grima Baja': 'GB', 'Pozo Gallo': 'PG', 'Agua Buena': 'AB',
    'Choro Bajo': 'VN', 'La Siberia': 'LS', 'Cantera': 'CT',
    'La Victoria': 'LV', 'El Raizal': 'RZ'
}

# --- ADJUSTED TITLES AND LABELS ---
variables = ['Mound_Area_Log', 'Shape_Ratio', 'Soil_pH', 'Light_Percentage', 'Temp_Diff']
# Professional titles using LaTeX for math formatting
professional_titles = [
    r'$\mathbf{Log_{10}(Mound\ Area\ [cm^2])}$', 
    'Shape Ratio', 
    'Soil pH', 
    'Canopy Openness (%)',
    'Thermal Regulation Capacity (°C)'
]
panel_labels = ['A', 'B', 'C', 'D', 'E']

def get_darker_color(hex_color, factor=0.4):
    rgb = mcolors.to_rgb(hex_color)
    darker_rgb = [max(0, x * (1 - factor)) for x in rgb]
    return mcolors.to_hex(darker_rgb)

line_palette = {k: get_darker_color(v) for k, v in species_palette.items()}

# Set style
sns.set_theme(style="ticks")
fig, axes = plt.subplots(3, 2, figsize=(7.5, 11))
axes = axes.flatten()

for i, var in enumerate(variables):
    unique_locations = sorted(df_final['Location'].unique())
    
    sns.boxplot(
        data=df_final, 
        x='Species', 
        y=var, 
        hue='Location', 
        hue_order=unique_locations,
        ax=axes[i],
        dodge=True,
        width=0.8,
        showfliers=False,
        linewidth=1.0
    )
    
    # Identify which species name corresponds to which X index (0 or 1)
    # This handles cases where 'C' might be first or 'L' might be first
    xtick_labels = [t.get_text() for t in axes[i].get_xticklabels()]

    # 1. Recolor the BOXES (Patches)
    for patch in axes[i].patches:
        x_mid = patch.get_path().get_extents().xmin + (patch.get_path().get_extents().width / 2)
        # Determine species based on which tick it's closer to
        spec_idx = 0 if x_mid < 0.5 else 1
        spec_key = xtick_labels[spec_idx]
        
        patch.set_facecolor(species_palette[spec_key])
        patch.set_edgecolor(line_palette[spec_key])
        patch.set_alpha(0.5)
        
    # 2. Recolor the LINES (Whiskers, Medians, Caps)
    # We look at the x-data of the line to decide its color
    for line in axes[i].lines:
        x_coords = line.get_xdata()
        x_avg = np.mean(x_coords)
        
        # Determine species based on x-coordinate
        spec_idx = 0 if x_avg < 0.5 else 1
        spec_key = xtick_labels[spec_idx]
        
        line.set_color(line_palette[spec_key])

    # Formatting
    axes[i].set_title(f'{var.replace("_", " ")}', fontsize=14, fontweight='bold')
    axes[i].set_ylabel('')
    axes[i].set_xlabel('Ant Species', fontsize=11)

    # --- ADD PANEL LABELS (A, B, C, D) ---
    axes[i].text(-0.15, 1.1, panel_labels[i], transform=axes[i].transAxes,
                 fontsize=14, fontweight='bold', va='top', ha='right')

    # Formatting with professional title
    axes[i].set_title(professional_titles[i], fontsize=12, fontweight='bold')
    axes[i].set_ylabel('')

    # --- CONDITIONAL X-AXIS TITLE ---
    # Only show 'Ant Species' for the first plot (index 0)
    if i == 0:
        axes[i].set_xlabel('Ant Species', fontsize=11)
    else:
        axes[i].set_xlabel('', fontsize=11)
    
    if axes[i].get_legend():
        axes[i].get_legend().remove()
    
    # Label Logic
    n_locs = len(unique_locations)
    width = 0.8
    offsets = np.linspace(-width/2 + width/(2*n_locs), width/2 - width/(2*n_locs), n_locs)
    
    y_min, y_max = axes[i].get_ylim()
    y_range = y_max - y_min
    axes[i].set_ylim(y_min - (y_range * 0.08), y_max)
    
    new_y_min, _ = axes[i].get_ylim()
    y_text = new_y_min + (y_range * 0.01)

    for j, spec in enumerate(xtick_labels):
        for k, loc in enumerate(unique_locations):
            x_pos = j + offsets[k]
            subset = df_final[(df_final['Species'] == spec) & (df_final['Location'] == loc)]
            
            if not subset.empty:
                code = location_codes.get(loc, loc)
                axes[i].text(
                    x_pos, y_text, code, 
                    ha='center', va='bottom', fontsize=7, 
                    fontweight='bold', color='#444444', rotation=90 
                )

# Hide the unused panel (Panel 6)
for j in range(len(variables), len(axes)):
    axes[j].axis('off')

# Custom Legend
legend_elements = [
    Line2D([0], [0], color=species_palette['C'], lw=6, label=r'$\mathit{A. \: cephalotes}$ (C)', alpha=0.5),
    Line2D([0], [0], color=species_palette['L'], lw=6, label=r'$\mathit{A. \: laevigata}$ (L)', alpha=0.5)
]

# Move the legend to the empty space (bottom right) or keep it centered
fig.legend(handles=legend_elements, loc='lower center', ncol=2, 
           bbox_to_anchor=(0.5, 0.02), frameon=False)

plt.tight_layout(rect=[0, 0.05, 1, 1])

# Save and Show
plt.savefig('Figure_Ant_Species_Distributions.jpg', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# ==============================================================================
# CREATE A RADAR PLOT OF QUANTITATIVE VARIABLES TO VISUALIZE THE REALIZED NICHE BREADTH
# ==============================================================================

# 1. Initialize and calculate metrics
dispersion_metrics = []

for var in array_variables:
    for species in df_final['Species'].unique():
        data = df_final[df_final['Species'] == species][var].dropna()
        
        # Calculate metrics
        std = data.std()
        mean = data.mean()
        cv = (std / mean) * 100 if mean != 0 else 0
        iqr = data.quantile(0.75) - data.quantile(0.25)
        
        dispersion_metrics.append({
            'Variable': var,
            'Species': species,
            'SD': std,
            'CV_%': cv,
            'IQR': iqr
        })

# 2. Create DataFrame and map professional names
df_dispersion = pd.DataFrame(dispersion_metrics)

variable_map = {
    'Mound_Area_m2': 'Mound area (m²)',
    'Shape_Ratio': 'Shape ratio',
    'Soil_pH': 'Soil pH',
    'Light_Percentage': 'Canopy openess (%)',
    'Temp_Diff': 'Thermal regulation capacity (°C)',
}
df_dispersion['Variable'] = df_dispersion['Variable'].map(variable_map)

# 3. Pivot the table for side-by-side species comparison
# This also handles potential duplicate entries by taking the first found
df_pivoted = df_dispersion.pivot_table(
    index='Variable', 
    columns='Species', 
    values=['SD', 'CV_%', 'IQR'],
    aggfunc='first'
)

# 4. Flatten columns and reorder for professional layout
df_pivoted.columns = [f'{metric} ({species})' for metric, species in df_pivoted.columns]

# Organizing columns: SD together, CV together, IQR together
final_cols = [
    'SD (C)', 'SD (L)', 
    'CV_% (C)', 'CV_% (L)', 
    'IQR (C)', 'IQR (L)'
]
df_final_table = df_pivoted[final_cols].round(3).reset_index()

# 5. Prepare Data for Spider Plot (Standardized Niche Breadth)
# We use SD as the base metric, standardized by the maximum SD for each variable
radar_df = df_pivoted[['SD (C)', 'SD (L)']].copy()

# 6. Calculate Relative Breadth (0 to 1 scale for each axis)
# This ensures "Apples vs Oranges" can be plotted on one spider web
for index, row in radar_df.iterrows():
    max_val = row.max()
    radar_df.loc[index, 'Rel_C'] = row['SD (C)'] / max_val
    radar_df.loc[index, 'Rel_L'] = row['SD (L)'] / max_val

categories = radar_df.index.tolist()
N = len(categories)

# Close the circle for the plot
values_c = radar_df['Rel_C'].tolist()
values_c += values_c[:1]
values_l = radar_df['Rel_L'].tolist()
values_l += values_l[:1]

angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]

# 7. Visualization - Standardized for Publication
fig = plt.figure(figsize=(5, 6)) # Increased size slightly to accommodate labels
ax = fig.add_subplot(111, polar=True)

# Draw one axe per variable + add labels
plt.xticks(angles[:-1], categories, color='#333333', size=12)

# Draw ylabels (Standardized Breadth)
ax.set_rlabel_position(0)
plt.yticks([0.25, 0.5, 0.75, 1.0], ["0.25", "0.50", "0.75", "1.0"], color="grey", size=8)
plt.ylim(0, 1.1)

# --- PANEL LABEL 'A' FIX ---
# Pinning the label to the absolute top-left of the figure frame
#fig.text(0.05, 0.925, 'A', fontsize=18, weight='bold', va='top', ha='left')

# Plot Species
# Plot Species A. cephalotes
ax.plot(angles, values_c, linewidth=1, linestyle='solid', 
        label=r'$\it{A. cephalotes}$', color=species_palette['C']) # Added 'r' before the string
# Plot Species A. laevigata
ax.plot(angles, values_l, linewidth=1, linestyle='dashed', 
        label=r'$\it{A. laevigata}$', color=species_palette['L']) # Added 'r' before the string
ax.fill(angles, values_c, species_palette['C'], alpha=0.1)
ax.fill(angles, values_l, species_palette['L'], alpha=0.1)

# Style adjustments
plt.title("Relative niche breadth", size=14, weight='bold')
plt.legend(loc='lower center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=False)

# --- THE BORDER FIX ---
# 1. Adjust subplots to leave space for labels so they don't push the border out
plt.subplots_adjust(left=0.15, right=0.85, top=0.85, bottom=0.15)

# Save with a small pad to ensure the 1px line is included
plt.savefig("Figure_Niche_SpiderPlot.png", dpi=300, bbox_inches='tight', pad_inches=0.1)
plt.savefig('Figure_Niche_SpiderPlot.svg', bbox_inches='tight')
plt.show()

# 5. Export and Display
print("--- Table S4: Dispersion Metrics and Niche Breadth ---")
print(df_final_table)

df_final_table.to_excel("Table_S4_Dispersion_Metrics.xlsx", index=False)
```

```python
# ==============================================================================
# PERFORM INDVAL ANALYSIS TO ESTIMATE AFFINITY OF SPECIES TO MACRO VEGETATION CATEGORIES
# ==============================================================================

import numpy as np
import pandas as pd

def run_indval_analysis(df, species_col, veg_col):
    # 1. Map to your specific definitions
    veg_mapping = {
        'G': 'Grasses',
        'B': 'Bushes',
        'T': 'Trees'
    }
    
    df_binary = df.copy()
    # Check for presence of each category (handles strings like 'GT' or 'GBT')
    for code, full_name in veg_mapping.items():
        df_binary[full_name] = df_binary[veg_col].apply(lambda x: 1 if code in str(x) else 0)

    species_list = df_binary[species_col].unique()
    attributes = list(veg_mapping.values())
    results = []

    for spec in species_list:
        # Standardize species names for publication
        spec_label = 'A. cephalotes' if spec == 'C' else 'A. laevigata'
        
        for attr in attributes:
            # --- Specificity (A) ---
            # Exclusivity: Probability that the site belongs to the species given the attribute
            n_sites_spec_attr = df_binary[(df_binary[species_col] == spec) & (df_binary[attr] == 1)].shape[0]
            n_sites_attr = df_binary[df_binary[attr] == 1].shape[0]
            A = n_sites_spec_attr / n_sites_attr if n_sites_attr > 0 else 0
            
            # --- Fidelity (B) ---
            # Probability of finding the attribute at a site occupied by the species
            n_sites_spec_total = df_binary[df_binary[species_col] == spec].shape[0]
            B = n_sites_spec_attr / n_sites_spec_total if n_sites_spec_total > 0 else 0
            
            # --- Indicator Value ---
            indval = np.sqrt(A * B)
            
            results.append({
                'Species': spec_label,
                'Vegetation Category': attr,
                'Specificity (A)': A,
                'Fidelity (B)': B,
                'IndVal Index': indval
            })
            
    return pd.DataFrame(results).sort_values(by=['Species', 'IndVal Index'], ascending=[True, False])

# Run analysis
results_vegetation = run_indval_analysis(df_final, 'Species', 'Vegetation')

# Final formatting for the table
df_indval_final = results_vegetation.round(3)
print("--- Table S6: Dispersion Metrics and Niche Breadth ---")
print(df_indval_final)

# Export
df_indval_final.to_excel("Table_S6_Vegetation_IndVal.xlsx", index=False)
```

```python
# ==============================================================================
# LOADINGS OF PC1 AND PC2
# Generate a supplementary table with the contribution of each variable to the PC1 and PC2
# ==============================================================================


# 1. Calculate squared loadings (contributions)
# Squaring the loadings gives the proportion of variance explained per variable
contributions = (pca.components_.T) ** 2

# 2. Create a DataFrame with the contributions
df_contributions = pd.DataFrame(
    contributions, columns=["PC1 (%)", "PC2 (%)"], index=features
)

# Multiply by 100 to convert proportions to percentages
df_contributions = df_contributions * 100

# 3. Reset the index and map the clean, formal variable names
df_contributions = df_contributions.reset_index().rename(
    columns={"index": "Variable"}
)
df_contributions["Variable"] = df_contributions["Variable"].map(
    variable_map_pca
)

# 4. Round the percentages to two decimal places for publication formatting
df_contributions["PC1 (%)"] = df_contributions["PC1 (%)"].round(2)
df_contributions["PC2 (%)"] = df_contributions["PC2 (%)"].round(2)

# 5. Sort by PC1 contribution descending to make the table easily readable
df_contributions = df_contributions.sort_values(by="PC1 (%)", ascending=False)

print("--- Table S5: Variable Contributions to PC1 and PC2 ---")
print(df_contributions.to_string(index=False))

# Export directly to your Excel workbook
df_contributions.to_excel("Table_S5_PCA_Variable_Contributions.xlsx", index=False)
```

```python
# ==============================================================================
# GLM ANALYSIS OF SOIL pH
# This analysis was used to evaluate any possible effect of the colony size
# on the soil pH.
# ==============================================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import seaborn as sns
import matplotlib.pyplot as plt


# ==========================================
# 1. DATA SIMULATION (Replace with your actual DataFrame)
# ==========================================

# Replace this section by loading your data file, for example:
# df = pd.read_csv("datos_fieldwork_charala.csv")

np.random.seed(42)


# ==========================================
# 2. GENERALIZED LINEAR MODEL (GLM)
# ==========================================

# We use the statsmodels formula interface (equivalent to the R formula syntax).
# 'family=sm.families.Gaussian()' indicates that we assume a normal distribution for soil pH.
# statsmodels automatically converts categorical variables (Species, Location) into dummy variables.

formula = "Soil_pH ~ Mound_Area + Shape_Ratio + Species + Location"

modelo = smf.glm(
    formula=formula,
    data=df_final,
    family=sm.families.Gaussian()
).fit()

# Print the statistical summary
print('--- Table S7 ----')
print(modelo.summary())


# ==========================================
# 3. GLM WITH INTERACTION
#    (Does the effect depend on species?)
# ==========================================

print("\n" + "=" * 50 + "\nMODEL WITH INTERACTION\n" + "=" * 50)

formula_interaccion = "Soil_pH ~ Mound_Area * Species + Shape_Ratio + Location"

modelo_int = smf.glm(
    formula=formula_interaccion,
    data=df_final,
    family=sm.families.Gaussian()
).fit()

print('--- Table S8 ----')
print(modelo_int.summary())


# ==========================================
# 1. EXPORT MODEL 1 (Additive) -> Table S7
# ==========================================

# Extract the parameter table (coefficients, p-values, and confidence intervals)
# as a DataFrame.

df_table_s7 = modelo.summary2().tables[1]

# Optional: Round the values for a cleaner publication format.

df_table_s7 = df_table_s7.round({
    'Coef.': 4,
    'Std.Err.': 4,
    'z': 3,
    'P>|z|': 4,
    '[0.025': 4,
    '0.975]': 4
})

# Export to Excel.

df_table_s7.to_excel(
    "Table_S7_Model1_Additive.xlsx",
    index=True
)


# ==========================================
# 2. EXPORT MODEL 2 (Interaction) -> Table S8
# ==========================================

# Extract the parameter table for the interaction model.

df_table_s8 = modelo_int.summary2().tables[1]

# Round values.

df_table_s8 = df_table_s8.round({
    'Coef.': 4,
    'Std.Err.': 4,
    'z': 3,
    'P>|z|': 4,
    '[0.025': 4,
    '0.975]': 4
})

# Export to Excel.

df_table_s8.to_excel(
    "Table_S8_Model2_Interaction.xlsx",
    index=True
)

print("Tables successfully exported to Excel!")


# ==========================================
# 4. VISUALIZATION OF THE CORRELATION
# ==========================================

# Scatter plot with separate trend lines for each species.

plt.figure(figsize=(8, 5))

sns.lmplot(
    x="Mound_Area",
    y="Soil_pH",
    hue="Species",
    data=df_final,
    palette="Set2",
    aspect=1.5,
    markers=["o", "s"]
)

plt.title("Effect of Mound Size on Soil pH")
plt.xlabel("Mound Area ($m^2$)")
plt.ylabel("Soil pH")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EXTRACT FROM ERA5 THE EVAPORATION FROM BARE SOIL OF THE YEARS 2024, 2025, AND 2026
# ERA5 VARIABLE evaporation_from_bare_soil
# Documentation: https://confluence.ecmwf.int/spaces/CKB/pages/140385202/ERA5-Land+data+documentation
# Because this code was created in August 2026, the values for 2026 were limited until July
# The  library cdsapi must be previously imported
# ============================================================

# Initialize CDS API client
client = cdsapi.Client()

# Define the bounding box:
# [North, West, South, East]
bbox = [8.0, -74.5, 6.0, -72.0]

# Years to download
years = ["2024", "2025", "2026"]

for year in years:

    # For complete years, request all 12 months.
    # For 2026, request only months that may already be available.
    if year == "2026":
        months = [f"{month:02d}" for month in range(1, 8)]  # January–July
    else:
        months = [f"{month:02d}" for month in range(1, 13)]

    request = {
        "variable": ["evaporation_from_bare_soil"],
        "year": year,
        "month": months,
        "day": [f"{day:02d}" for day in range(1, 32)],
        "time": [f"{hour:02d}:00" for hour in range(24)],
        "area": bbox,
        "data_format": "grib",
        "download_format": "unarchived",
    }

    output_file = f"era5_land_bare_soil_evaporation_{year}.grib"

    print(f"Downloading {year}...")

    client.retrieve(
        "reanalysis-era5-land",
        request
    ).download(output_file)

    print(f"Saved: {output_file}")
```

```python
# ============================================================
# THREE-PANEL ERA5-LAND TEMPERATURE MAP
# FEBRUARY 2024, 2025, AND 2026
# ============================================================

import xarray as xr
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib as mpl
import contextily as cx

from matplotlib_scalebar.scalebar import ScaleBar


# ============================================================
# 1. INPUT FILES AND GLOBAL SETTINGS
# ============================================================

temperature_files = {
    2024: "../data/era5_land/era5_land_2m_temperature_february_2024.grib",
    2025: "../data/era5_land/era5_land_2m_temperature_february_2025.grib",
    2026: "../data/era5_land/era5_land_2m_temperature_february_2026.grib"
}

copernicus_variable = "t2m"

# Fixed temperature range for all panels.
# Adjust these values if necessary after checking the data.
temperature_min = 10
temperature_max = 30

# Map appearance
mpl.rcParams["hatch.linewidth"] = 0.5


# ============================================================
# 2. LOAD SAMPLING LOCATIONS
# ============================================================

df_coordinates = pd.read_csv(
    DATA_FIELDWORK_LOCATION_COORDINATES
)

# Include all sampling locations
df_coordinates_map = df_coordinates.copy()

# Convert geographic coordinates to a GeoDataFrame
gdf = gpd.GeoDataFrame(
    df_coordinates_map,
    geometry=gpd.points_from_xy(
        df_coordinates_map.Longitude,
        df_coordinates_map.Latitude
    ),
    crs="EPSG:4326"
)

# Reproject to Web Mercator because OpenStreetMap tiles use EPSG:3857
gdf = gdf.to_crs(epsg=3857)


# ============================================================
# 3. DEFINE A COMMON MAP EXTENT FOR ALL PANELS
# ============================================================

x_min, y_min, x_max, y_max = gdf.total_bounds

margin = 1200

map_xlim = (
    x_min - margin * 3.5,
    x_max + margin * 1.5
)

map_ylim = (
    y_min - margin,
    y_max + margin
)


# ============================================================
# 4. FUNCTION TO LOAD AND PROCESS ERA5-LAND TEMPERATURE DATA
# ============================================================

def load_temperature_raster(grib_file):
    """
    Load ERA5-Land 2 m temperature data from a GRIB file,
    calculate the mean across time and forecast-step dimensions,
    reproject the raster to Web Mercator, and convert Kelvin to Celsius.
    """

    # Open the GRIB dataset using cfgrib
    ds = xr.open_dataset(
        grib_file,
        engine="cfgrib"
    )

    # Extract the 2 m temperature variable
    data_array = ds[copernicus_variable]

    # Average across temporal dimensions when present
    dimensions_to_average = [
        dim for dim in ["time", "step"]
        if dim in data_array.dims
    ]

    if dimensions_to_average:
        temp_mean = data_array.mean(
            dim=dimensions_to_average
        )
    else:
        temp_mean = data_array

    # Remove dimensions of length one
    temp_final = temp_mean.squeeze()

    # ERA5-Land geographic coordinates are WGS84
    temp_final = temp_final.rio.write_crs(
        "EPSG:4326"
    )

    # Reproject to Web Mercator to match the basemap
    temp_raster = temp_final.rio.reproject(
        "EPSG:3857"
    )

    # Convert temperature from Kelvin to Celsius
    temp_raster_c = temp_raster - 273.15

    ds.close()

    return temp_raster_c


# ============================================================
# 5. LOAD ALL THREE TEMPERATURE RASTERS
# ============================================================

temperature_rasters = {
    year: load_temperature_raster(file_path)
    for year, file_path in temperature_files.items()
}


# ============================================================
# 6. CROP RASTERS TO THE REGION DISPLAYED IN THE MAP
# ============================================================

def crop_raster_to_map_extent(raster):
    """
    Crop a raster to the geographic extent displayed in the map.
    """

    # Handle either ascending or descending y coordinates
    if raster.y[0] > raster.y[-1]:
        y_slice = slice(
            map_ylim[1],
            map_ylim[0]
        )
    else:
        y_slice = slice(
            map_ylim[0],
            map_ylim[1]
        )

    return raster.sel(
        x=slice(
            map_xlim[0],
            map_xlim[1]
        ),
        y=y_slice
    )


temperature_cropped = {
    year: crop_raster_to_map_extent(raster)
    for year, raster in temperature_rasters.items()
}

# ============================================================
# 7. DEFINE A COMMON TEMPERATURE COLOR SCALE
# ============================================================

temperature_min = min(
    float(temperature_cropped[2024].min()),
    float(temperature_cropped[2025].min()),
    float(temperature_cropped[2026].min())
)

temperature_max = max(
    float(temperature_cropped[2024].max()),
    float(temperature_cropped[2025].max()),
    float(temperature_cropped[2026].max())
)

print(
    f"Shared temperature scale: "
    f"{temperature_min:.2f} to {temperature_max:.2f} °C"
)
    
# ============================================================
# 8. FUNCTION TO PLOT ONE TEMPERATURE PANEL
# ============================================================

def plot_temperature_panel(
    ax,
    temperature_raster,
    year
):
    """
    Plot one ERA5-Land temperature raster together with
    sampling locations and an OpenStreetMap basemap.
    """

    image = temperature_raster.plot.imshow(
        ax=ax,
        x="x",
        y="y",
        add_colorbar=False,
        cmap="YlOrRd",
        alpha=0.80,
        zorder=2,
        vmin=temperature_min,
        vmax=temperature_max
    )

    # Add sampling locations
    gdf.plot(
        ax=ax,
        color="black",
        markersize=35,
        edgecolor="white",
        linewidth=0.8,
        zorder=5
    )

    # Add location labels
    label_offset = 500

    for x, y, label in zip(
        gdf.geometry.x,
        gdf.geometry.y,
        gdf["Location"]
    ):

        ax.text(
            x,
            y + label_offset,
            str(label),
            fontsize=5.5,
            fontweight="bold",
            ha="center",
            va="bottom",
            zorder=6,
            bbox=dict(
                facecolor="white",
                alpha=0.6,
                edgecolor="none",
                pad=1
            )
        )

    # Apply identical geographic extent to all panels
    ax.set_xlim(map_xlim)
    ax.set_ylim(map_ylim)

    # Add OpenStreetMap basemap underneath the ERA5 raster
    cx.add_basemap(
        ax,
        source=cx.providers.OpenStreetMap.Mapnik,
        zoom=14,
        attribution="",
        zorder=0
    )

    # Add scale bar
    scalebar = ScaleBar(
        dx=1,
        units="m",
        location="lower right",
        frameon=True,
        color="black",
        box_alpha=0.5,
        rotation="vertical",
        font_properties={
            "size": 7,
            "weight": "bold"
        }
    )

    ax.add_artist(scalebar)

    # Add panel title
    ax.set_title(
        f"Average temperature\nFebruary {year}",
        fontsize=11,
        fontweight="bold",
        pad=8
    )

    # Preserve geographic proportions
    ax.set_aspect("equal")

    # Remove axis ticks and borders
    ax.set_axis_off()

    return image

# ============================================================
# 9. CREATE THREE-PANEL FIGURE
# ============================================================

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(7, 8),
    dpi=300,
    constrained_layout=False
)


# ============================================================
# 10. PLOT 2024, 2025, AND 2026
# ============================================================

years = [
    2024,
    2025,
    2026
]

images = []

for ax, year in zip(
    axes,
    years
):

    image = plot_temperature_panel(
        ax=ax,
        temperature_raster=temperature_rasters[year],
        year=year
    )

    images.append(image)


# ============================================================
# 11. ADD A SHARED TEMPERATURE COLORBAR
# ============================================================

# All panels use the same temperature range and colormap,
# therefore a single shared colorbar is used.

cbar = fig.colorbar(
    images[0],
    ax=axes,
    orientation="horizontal",
    fraction=0.04,
    pad=0.11,
    aspect=45
)

cbar.set_label(
    "2 m temperature (°C)",
    fontsize=11,
    fontweight="bold"
)

cbar.ax.tick_params(
    labelsize=9
)


# ============================================================
# 12. FINAL FIGURE FORMATTING
# ============================================================

fig.subplots_adjust(
    left=0.04,
    right=0.98,
    top=0.88,
    bottom=0.18,
    wspace=0.015
)


# ============================================================
# 13. SAVE FIGURE
# ============================================================
plt.savefig(
    "S3_Temperature_February_2024_2025_2026.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "S3_Temperature_February_2024_2025_2026.svg",
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 14. MAP ATTRIBUTION
# ============================================================

print("--- Map Attribution ---")
print("Base map: © OpenStreetMap contributors")
print(cx.providers.OpenStreetMap.Mapnik)
```

```python
# ============================================================
# EXTRACT TEMPERATURE DIRECTLY FROM ORIGINAL ERA5-LAND GRID
# ============================================================

temperature_at_locations = []

for _, row in df_coordinates.iterrows():

    location_name = row["Location"]
    latitude = row["Latitude"]
    longitude = row["Longitude"]

    record = {
        "Location": location_name,
        "Latitude": latitude,
        "Longitude": longitude
    }

    for year, grib_file in temperature_files.items():

        ds = xr.open_dataset(
            grib_file,
            engine="cfgrib"
        )

        t2m = ds[copernicus_variable]

        # Calculate February mean
        dimensions_to_average = [
            dim for dim in ["time", "step"]
            if dim in t2m.dims
        ]

        if dimensions_to_average:
            t2m_mean = t2m.mean(
                dim=dimensions_to_average
            )
        else:
            t2m_mean = t2m

        # Select nearest ERA5-Land grid cell
        site_temperature = t2m_mean.sel(
            latitude=latitude,
            longitude=longitude,
            method="nearest"
        )

        # Convert Kelvin to Celsius
        temperature_c = (
            float(site_temperature.values) - 273.15
        )

        record[f"February_{year}_C"] = temperature_c

        ds.close()

    temperature_at_locations.append(record)


df_temperature_locations = pd.DataFrame(
    temperature_at_locations
)

df_temperature_locations = df_temperature_locations.sort_values(
    by="February_2024_C",
    ascending=False
)

print(
    df_temperature_locations.to_string(
        index=False,
        float_format=lambda x: f"{x:.2f}"
    )
)
```

```python
# ============================================================
# THREE-PANEL ERA5-LAND TOTAL PRECIPITATION MAP
# 2024, 2025, AND JANUARY-JULY 2026
# ============================================================

import xarray as xr
import rioxarray
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as cx

from matplotlib_scalebar.scalebar import ScaleBar


# ============================================================
# 1. INPUT FILES AND GLOBAL SETTINGS
# ============================================================

precipitation_files = {
    2024: "../data/era5_land/era5_land_total_precipitation_2024.grib",
    2025: "../data/era5_land/era5_land_total_precipitation_2025.grib",
    2026: "../data/era5_land/ra5_land_total_precipitation_2026.grib"
}

# ERA5-Land total precipitation variable
copernicus_variable = "tp"


# ============================================================
# 2. LOAD SAMPLING LOCATIONS
# ============================================================

df_coordinates = pd.read_csv(
    DATA_FIELDWORK_LOCATION_COORDINATES
)

# Include all sampling locations
df_coordinates_map = df_coordinates.copy()

# Convert geographic coordinates to a GeoDataFrame
gdf = gpd.GeoDataFrame(
    df_coordinates_map,
    geometry=gpd.points_from_xy(
        df_coordinates_map.Longitude,
        df_coordinates_map.Latitude
    ),
    crs="EPSG:4326"
)

# Reproject to Web Mercator because OpenStreetMap tiles use EPSG:3857
gdf = gdf.to_crs(epsg=3857)


# ============================================================
# 3. DEFINE A COMMON MAP EXTENT FOR ALL PANELS
# ============================================================

x_min, y_min, x_max, y_max = gdf.total_bounds

margin = 1200

map_xlim = (
    x_min - margin * 3.5,
    x_max + margin * 1.5
)

map_ylim = (
    y_min - margin,
    y_max + margin
)


# ============================================================
# 4. FUNCTION TO LOAD AND PROCESS ERA5-LAND PRECIPITATION
# ============================================================

def load_precipitation_raster(grib_file):
    """
    Load ERA5-Land total precipitation from a GRIB file,
    extract the 24-hour accumulated value for each day,
    sum daily precipitation across the available period,
    and reproject the resulting raster to Web Mercator.

    ERA5-Land total precipitation is expressed in meters
    of water equivalent.
    """

    # Open the GRIB dataset
    ds = xr.open_dataset(
        grib_file,
        engine="cfgrib"
    )

    # Extract total precipitation
    data_array = ds[copernicus_variable]

    print(
        f"Loading {grib_file}\n"
        f"Dimensions: {data_array.dims}"
    )

    # Select the final forecast step, corresponding to the
    # accumulated precipitation over each 24-hour period
    if "step" in data_array.dims:
        daily_totals = data_array.sel(
            step="24 hours"
        )
    else:
        daily_totals = data_array

    # Sum daily totals across the available dates
    if "time" in daily_totals.dims:
        precipitation_total = daily_totals.sum(
            dim="time"
        )
    else:
        precipitation_total = daily_totals

    # Remove dimensions of length one
    precipitation_final = precipitation_total.squeeze()

    # Assign the original geographic CRS
    precipitation_final = precipitation_final.rio.write_crs(
        "EPSG:4326"
    )

    # Reproject to Web Mercator to match OpenStreetMap
    precipitation_raster = precipitation_final.rio.reproject(
        "EPSG:3857"
    )

    ds.close()

    return precipitation_raster



# ============================================================
# 5. LOAD ALL THREE PRECIPITATION RASTERS
# ============================================================

precipitation_rasters = {
    year: load_precipitation_raster(file_path)
    for year, file_path in precipitation_files.items()
}


# ============================================================
# 6. FUNCTION TO CROP ONLY THE REGION OF THE MAP TO BE PLOTTED
# FROM THE BROAD REGION OF THE ERA5 DATASET THAT WE DOWNLOADED
# THIS IS IMPORTANT TO AVOID ESTIMATING MAX VALUES FROM A REGION
# THAT IS NOT PLOTTED, THUS, VISUALIZATION CAN BE DIFFICULT TO INTERPRET
# ============================================================
def crop_raster_to_map_extent(raster):
    """
    Crop a raster to the geographic extent displayed in the map.
    """

    # Handle either ascending or descending y coordinates
    if raster.y[0] > raster.y[-1]:
        y_slice = slice(map_ylim[1], map_ylim[0])
    else:
        y_slice = slice(map_ylim[0], map_ylim[1])

    return raster.sel(
        x=slice(map_xlim[0], map_xlim[1]),
        y=y_slice
    )

# The estimated total cumulative precitation is cropeed to the map that is actually plotted
precipitation_cropped = {
    year: crop_raster_to_map_extent(raster)
    for year, raster in precipitation_rasters.items()
}


# ============================================================
# 7. DEFINE COLOR SCALES
# ============================================================

# 2024 and 2025 represent complete calendar years and therefore
# share the same color scale for direct comparison.

annual_precipitation_min = min(
    float(precipitation_cropped[2024].min()),
    float(precipitation_cropped[2025].min())
)


annual_precipitation_max = max(
    float(precipitation_cropped[2024].max()),
    float(precipitation_cropped[2025].max())
)

# 2026 includes only January-July and therefore uses its own
# color scale rather than being directly scaled against full years.

partial_2026_min = float(
    precipitation_cropped[2026].min()
)

partial_2026_max = float(
    precipitation_cropped[2026].max()
)

print(
    f"2024-2025 shared precipitation scale: "
    f" {annual_precipitation_min:.3f} to {annual_precipitation_max:.3f} m"
)

print(
    f"2026 precipitation scale: "
    f"{partial_2026_min:.3f} to {partial_2026_max:.3f} m"
)


# ============================================================
# 8. FUNCTION TO PLOT ONE PRECIPITATION PANEL
# ============================================================

def plot_precipitation_panel(
    ax,
    precipitation_raster,
    title,
    vmin,
    vmax
):
    """
    Plot one cumulative precipitation raster together with
    sampling locations and an OpenStreetMap basemap.
    """

    # Plot precipitation raster
    image = precipitation_raster.plot.imshow(
        ax=ax,
        x="x",
        y="y",
        add_colorbar=False,
        cmap="YlGnBu",
        alpha=0.80,
        zorder=2,
        vmin=vmin,
        vmax=vmax
    )

    # Add sampling locations
    gdf.plot(
        ax=ax,
        color="black",
        markersize=35,
        edgecolor="white",
        linewidth=0.8,
        zorder=5
    )

    # Add location labels
    label_offset = 500

    for x, y, label in zip(
        gdf.geometry.x,
        gdf.geometry.y,
        gdf["Location"]
    ):

        ax.text(
            x,
            y + label_offset,
            str(label),
            fontsize=5.5,
            fontweight="bold",
            ha="center",
            va="bottom",
            zorder=6,
            bbox=dict(
                facecolor="white",
                alpha=0.6,
                edgecolor="none",
                pad=1
            )
        )

    # Apply identical geographic extent to all panels
    ax.set_xlim(map_xlim)
    ax.set_ylim(map_ylim)

    # Add OpenStreetMap underneath the precipitation raster
    cx.add_basemap(
        ax,
        source=cx.providers.OpenStreetMap.Mapnik,
        zoom=14,
        attribution="",
        zorder=0
    )

    # Add scale bar
    scalebar = ScaleBar(
        dx=1,
        units="m",
        location="lower right",
        frameon=True,
        color="black",
        box_alpha=0.5,
        rotation="vertical",
        font_properties={
            "size": 7,
            "weight": "bold"
        }
    )

    ax.add_artist(scalebar)

    # Add panel title
    ax.set_title(
        title,
        fontsize=11,
        fontweight="bold",
        pad=8
    )

    # Preserve geographic proportions
    ax.set_aspect("equal")

    # Remove axis ticks and borders
    ax.set_axis_off()

    return image


# ============================================================
# 9. CREATE THREE-PANEL FIGURE
# ============================================================

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(7, 8),
    dpi=300,
    constrained_layout=False
)



# ============================================================
# 10. PLOT 2024, 2025, AND 2026
# ============================================================

image_2024 = plot_precipitation_panel(
    ax=axes[0],
    precipitation_raster=precipitation_rasters[2024],
    title="Total precipitation\n2024",
    vmin=annual_precipitation_min,
    vmax=annual_precipitation_max
)

image_2025 = plot_precipitation_panel(
    ax=axes[1],
    precipitation_raster=precipitation_rasters[2025],
    title="Total precipitation\n2025",
    vmin=annual_precipitation_min,
    vmax=annual_precipitation_max
)

image_2026 = plot_precipitation_panel(
    ax=axes[2],
    precipitation_raster=precipitation_rasters[2026],
    title="Cumulative precipitation\nJan-Jul 2026",
    vmin=partial_2026_min,
    vmax=partial_2026_max
)


# ============================================================
# 11. ADJUST PANEL SPACING
# ============================================================

fig.subplots_adjust(
    left=0.04,
    right=0.98,
    top=0.88,
    bottom=0.24,
    wspace=0.015
)

# Render once so the final axis positions are available
fig.canvas.draw()


# ============================================================
# 12. ADD SHARED COLORBAR FOR 2024 AND 2025
# ============================================================

# Obtain the positions of the first two panels
bbox_2024 = axes[0].get_position()
bbox_2025 = axes[1].get_position()

# Position the shared colorbar below panels 2024 and 2025
annual_cbar_ax = fig.add_axes([
    bbox_2024.x0,
    bbox_2024.y0 - 0.085,
    bbox_2025.x1 - bbox_2024.x0,
    0.018
])

annual_cbar = fig.colorbar(
    image_2024,
    cax=annual_cbar_ax,
    orientation="horizontal"
)

annual_cbar.set_label(
    "Total annual precipitation (m)",
    fontsize=10,
    fontweight="bold"
)

annual_cbar.ax.tick_params(
    labelsize=8
)


# ============================================================
# 13. ADD SEPARATE COLORBAR FOR JANUARY-JULY 2026
# ============================================================

bbox_2026 = axes[2].get_position()

# Position the 2026 colorbar below the third panel
partial_cbar_ax = fig.add_axes([
    bbox_2026.x0,
    bbox_2026.y0 - 0.085,
    bbox_2026.width,
    0.018
])

partial_cbar = fig.colorbar(
    image_2026,
    cax=partial_cbar_ax,
    orientation="horizontal"
)

partial_cbar.set_label(
    "Jan-Jul precipitation (m)",
    fontsize=10,
    fontweight="bold"
)

partial_cbar.ax.tick_params(
    labelsize=8
)


# ============================================================
# 14. SAVE FIGURE
# ============================================================

plt.savefig(
    "S4_Total_Precipitation_2024_2025_2026.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "S4_Total_Precipitation_2024_2025_2026.svg",
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 15. MAP ATTRIBUTION
# ============================================================

print("--- Map Attribution ---")
print("Base map: © OpenStreetMap contributors")
print(cx.providers.OpenStreetMap.Mapnik)
```

```python
# ============================================================
# EXTRACT CUMULATIVE PRECIPITATION AT SAMPLING LOCATIONS
# ============================================================

precipitation_at_locations = []

for _, row in df_coordinates.iterrows():

    location_name = row["Location"]
    latitude = row["Latitude"]
    longitude = row["Longitude"]

    record = {
        "Location": location_name,
        "Latitude": latitude,
        "Longitude": longitude
    }

    for year, grib_file in precipitation_files.items():

        # Open original ERA5-Land GRIB file
        ds = xr.open_dataset(
            grib_file,
            engine="cfgrib"
        )

        # Extract total precipitation
        tp = ds["tp"]

        # Select the 24-hour accumulation for each day
        if "step" in tp.dims:
            daily_totals = tp.sel(
                step="24 hours"
            )
        else:
            daily_totals = tp

        # Sum daily precipitation across the available period
        if "time" in daily_totals.dims:
            precipitation_total = daily_totals.sum(
                dim="time"
            )
        else:
            precipitation_total = daily_totals

        precipitation_total = precipitation_total.squeeze()

        # Extract the nearest ERA5-Land grid cell
        site_precipitation = precipitation_total.sel(
            latitude=latitude,
            longitude=longitude,
            method="nearest"
        )

        # ERA5-Land total precipitation is already expressed in meters
        precipitation_m = float(
            site_precipitation.values
        )

        record[f"Precipitation_{year}_m"] = precipitation_m

        ds.close()

    precipitation_at_locations.append(record)


# Convert results to a DataFrame
df_precipitation_locations = pd.DataFrame(
    precipitation_at_locations
)

df_precipitation_locations = df_precipitation_locations.sort_values(
    by="Precipitation_2024_m",
    ascending=False
)


# Print results
print("\nCumulative precipitation at sampling locations:")

print(
    df_precipitation_locations.to_string(
        index=False,
        float_format=lambda x: f"{x:.3f}"
    )
)
```

```python
# ============================================================
# THREE-PANEL ERA5-LAND BARE-SOIL EVAPORATION MAP
# 2024, 2025, AND JANUARY-JULY 2026
# ============================================================

import xarray as xr
import rioxarray
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as cx

from matplotlib_scalebar.scalebar import ScaleBar


# ============================================================
# 1. INPUT FILES AND GLOBAL SETTINGS
# ============================================================

evaporation_files = {
    2024: "../data/era5_land/era5_land_bare_soil_evaporation_2024.grib",
    2025: "../data/era5_land/era5_land_bare_soil_evaporation_2025.grib",
    2026: "../data/era5_land/era5_land_bare_soil_evaporation_2026.grib"
}

# ERA5-Land evaporation from bare soil
copernicus_variable = "evabs"


# ============================================================
# 2. LOAD SAMPLING LOCATIONS
# ============================================================

df_coordinates = pd.read_csv(
    DATA_FIELDWORK_LOCATION_COORDINATES
)

# Include all sampling locations
df_coordinates_map = df_coordinates.copy()

# Convert geographic coordinates to a GeoDataFrame
gdf = gpd.GeoDataFrame(
    df_coordinates_map,
    geometry=gpd.points_from_xy(
        df_coordinates_map.Longitude,
        df_coordinates_map.Latitude
    ),
    crs="EPSG:4326"
)

# Reproject to Web Mercator because OpenStreetMap tiles use EPSG:3857
gdf = gdf.to_crs(epsg=3857)


# ============================================================
# 3. DEFINE A COMMON MAP EXTENT FOR ALL PANELS
# ============================================================

x_min, y_min, x_max, y_max = gdf.total_bounds

margin = 1200

map_xlim = (
    x_min - margin * 3.5,
    x_max + margin * 1.5
)

map_ylim = (
    y_min - margin,
    y_max + margin
)


# ============================================================
# 4. FUNCTION TO LOAD AND PROCESS ERA5-LAND EVAPORATION
# ============================================================

def load_evaporation_raster(grib_file):
    """
    Load ERA5-Land evaporation from bare soil from a GRIB file,
    extract the 24-hour accumulated value for each day,
    sum daily evaporation across the available period,
    convert the accumulated values from meters to millimeters,
    and reproject the resulting raster to Web Mercator.

    ERA5-Land evaporation is conventionally represented as a
    negative flux. Absolute values are therefore used here to
    represent cumulative evaporation magnitude as positive values.
    """

    # Open the GRIB dataset
    ds = xr.open_dataset(
        grib_file,
        engine="cfgrib"
    )

    # Extract evaporation from bare soil
    data_array = ds[copernicus_variable]

    print(
        f"Loading {grib_file}\n"
        f"Dimensions: {data_array.dims}"
    )

    # Select the final forecast step, corresponding to the
    # accumulated evaporation over each 24-hour period
    if "step" in data_array.dims:
        try:
            daily_totals = data_array.sel(
                step="24 hours"
            )
        except KeyError:
            # Fallback to the final available accumulation step
            daily_totals = data_array.isel(
                step=-1
            )
    else:
        daily_totals = data_array

    # Sum daily evaporation across the available dates
    if "time" in daily_totals.dims:
        evaporation_total = daily_totals.sum(
            dim="time"
        )
    else:
        evaporation_total = daily_totals

    # Convert negative evaporation values to positive magnitudes
    # and convert meters of water equivalent to millimeters
    evaporation_total_mm = abs(
        evaporation_total
    ) * 1000

    # Remove dimensions of length one
    evaporation_final = evaporation_total_mm.squeeze()

    # Assign the original geographic CRS
    evaporation_final = evaporation_final.rio.write_crs(
        "EPSG:4326"
    )

    # Reproject to Web Mercator to match OpenStreetMap
    evaporation_raster = evaporation_final.rio.reproject(
        "EPSG:3857"
    )

    ds.close()

    return evaporation_raster


# ============================================================
# 5. LOAD ALL THREE EVAPORATION RASTERS
# ============================================================

evaporation_rasters = {
    year: load_evaporation_raster(file_path)
    for year, file_path in evaporation_files.items()
}


# ============================================================
# 6. CROP RASTERS TO THE REGION DISPLAYED IN THE MAP
# ============================================================

def crop_raster_to_map_extent(raster):
    """
    Crop a raster to the geographic extent displayed in the map.
    """

    # Handle either ascending or descending y coordinates
    if raster.y[0] > raster.y[-1]:
        y_slice = slice(
            map_ylim[1],
            map_ylim[0]
        )
    else:
        y_slice = slice(
            map_ylim[0],
            map_ylim[1]
        )

    return raster.sel(
        x=slice(
            map_xlim[0],
            map_xlim[1]
        ),
        y=y_slice
    )


# Crop cumulative evaporation to the region that is actually plotted.
# This prevents values outside the study-area view from determining
# the color scale used in the figure.
evaporation_cropped = {
    year: crop_raster_to_map_extent(raster)
    for year, raster in evaporation_rasters.items()
}


# ============================================================
# 7. DEFINE COLOR SCALES
# ============================================================

# 2024 and 2025 represent complete calendar years and therefore
# share the same color scale for direct comparison.

annual_evaporation_min = min(
    float(evaporation_cropped[2024].min()),
    float(evaporation_cropped[2025].min())
)

annual_evaporation_max = max(
    float(evaporation_cropped[2024].max()),
    float(evaporation_cropped[2025].max())
)

# 2026 includes only January-July and therefore uses its own
# color scale rather than being directly scaled against full years.

partial_2026_min = float(
    evaporation_cropped[2026].min()
)

partial_2026_max = float(
    evaporation_cropped[2026].max()
)

print(
    f"2024-2025 shared evaporation scale: "
    f"{annual_evaporation_min:.1f} to "
    f"{annual_evaporation_max:.1f} mm"
)

print(
    f"2026 evaporation scale: "
    f"{partial_2026_min:.1f} to "
    f"{partial_2026_max:.1f} mm"
)


# ============================================================
# 8. FUNCTION TO PLOT ONE EVAPORATION PANEL
# ============================================================

def plot_evaporation_panel(
    ax,
    evaporation_raster,
    title,
    vmin,
    vmax
):
    """
    Plot one cumulative bare-soil evaporation raster together
    with sampling locations and an OpenStreetMap basemap.
    """

    # Plot evaporation raster
    image = evaporation_raster.plot.imshow(
        ax=ax,
        x="x",
        y="y",
        add_colorbar=False,
        cmap="YlGnBu",
        alpha=0.80,
        zorder=2,
        vmin=vmin,
        vmax=vmax
    )

    # Add sampling locations
    gdf.plot(
        ax=ax,
        color="black",
        markersize=35,
        edgecolor="white",
        linewidth=0.8,
        zorder=5
    )

    # Add location labels
    label_offset = 500

    for x, y, label in zip(
        gdf.geometry.x,
        gdf.geometry.y,
        gdf["Location"]
    ):

        ax.text(
            x,
            y + label_offset,
            str(label),
            fontsize=5.5,
            fontweight="bold",
            ha="center",
            va="bottom",
            zorder=6,
            bbox=dict(
                facecolor="white",
                alpha=0.6,
                edgecolor="none",
                pad=1
            )
        )

    # Apply identical geographic extent to all panels
    ax.set_xlim(map_xlim)
    ax.set_ylim(map_ylim)

    # Add OpenStreetMap underneath the evaporation raster
    cx.add_basemap(
        ax,
        source=cx.providers.OpenStreetMap.Mapnik,
        zoom=14,
        attribution="",
        zorder=0
    )

    # Add scale bar
    scalebar = ScaleBar(
        dx=1,
        units="m",
        location="lower right",
        frameon=True,
        color="black",
        box_alpha=0.5,
        rotation="vertical",
        font_properties={
            "size": 7,
            "weight": "bold"
        }
    )

    ax.add_artist(scalebar)

    # Add panel title
    ax.set_title(
        title,
        fontsize=11,
        fontweight="bold",
        pad=8
    )

    # Preserve geographic proportions
    ax.set_aspect("equal")

    # Remove axis ticks and borders
    ax.set_axis_off()

    return image


# ============================================================
# 9. CREATE THREE-PANEL FIGURE
# ============================================================

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(7, 8),
    dpi=300,
    constrained_layout=False
)


# ============================================================
# 10. PLOT 2024, 2025, AND 2026
# ============================================================

image_2024 = plot_evaporation_panel(
    ax=axes[0],
    evaporation_raster=evaporation_rasters[2024],
    title="Total bare-soil\nevaporation 2024", 
    vmin=annual_evaporation_min,
    vmax=annual_evaporation_max
)

image_2025 = plot_evaporation_panel(
    ax=axes[1],
    evaporation_raster=evaporation_rasters[2025],
    title="Total bare-soil\nevaporation 2025",
    vmin=annual_evaporation_min,
    vmax=annual_evaporation_max
)

image_2026 = plot_evaporation_panel(
    ax=axes[2],
    evaporation_raster=evaporation_rasters[2026],
    title="Cumulative bare-soil\nevaporationn Jan-Jul 2026",
    vmin=partial_2026_min,
    vmax=partial_2026_max
)


# ============================================================
# 11. ADJUST PANEL SPACING
# ============================================================

fig.subplots_adjust(
    left=0.04,
    right=0.98,
    top=0.88,
    bottom=0.24,
    wspace=0.015
)

# Render once so the final axis positions are available
fig.canvas.draw()


# ============================================================
# 12. ADD SHARED COLORBAR FOR 2024 AND 2025
# ============================================================

# Obtain the positions of the first two panels
bbox_2024 = axes[0].get_position()
bbox_2025 = axes[1].get_position()

# Position the shared colorbar below panels 2024 and 2025
annual_cbar_ax = fig.add_axes([
    bbox_2024.x0,
    bbox_2024.y0 - 0.085,
    bbox_2025.x1 - bbox_2024.x0,
    0.018
])

annual_cbar = fig.colorbar(
    image_2024,
    cax=annual_cbar_ax,
    orientation="horizontal"
)

annual_cbar.set_label(
    "Total annual bare-soil evaporation (mm)",
    fontsize=10,
    fontweight="bold"
)

annual_cbar.ax.tick_params(
    labelsize=8
)


# ============================================================
# 13. ADD SEPARATE COLORBAR FOR JANUARY-JULY 2026
# ============================================================

bbox_2026 = axes[2].get_position()

# Position the 2026 colorbar below the third panel
partial_cbar_ax = fig.add_axes([
    bbox_2026.x0,
    bbox_2026.y0 - 0.085,
    bbox_2026.width,
    0.018
])

partial_cbar = fig.colorbar(
    image_2026,
    cax=partial_cbar_ax,
    orientation="horizontal"
)

partial_cbar.set_label(
    "Jan-Jul bare-soil evaporation (mm)",
    fontsize=10,
    fontweight="bold"
)

partial_cbar.ax.tick_params(
    labelsize=8
)


# ============================================================
# 14. SAVE FIGURE
# ============================================================

plt.savefig(
    "S5_Bare_Soil_Evaporation_2024_2025_2026.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "S5_Bare_Soil_Evaporation_2024_2025_2026.svg",
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 15. MAP ATTRIBUTION
# ============================================================

print("--- Map Attribution ---")
print("Base map: © OpenStreetMap contributors")
print(cx.providers.OpenStreetMap.Mapnik)
```

```python
# ============================================================
# EXTRACT CUMULATIVE BARE-SOIL EVAPORATION AT SAMPLING LOCATIONS
# ============================================================

evaporation_at_locations = []

for _, row in df_coordinates.iterrows():

    location_name = row["Location"]
    latitude = row["Latitude"]
    longitude = row["Longitude"]

    record = {
        "Location": location_name,
        "Latitude": latitude,
        "Longitude": longitude
    }

    for year, grib_file in evaporation_files.items():

        # Open original ERA5-Land GRIB file
        ds = xr.open_dataset(
            grib_file,
            engine="cfgrib"
        )

        # Extract evaporation from bare soil
        evabs = ds["evabs"]

        # Select the 24-hour accumulation for each day
        if "step" in evabs.dims:
            try:
                daily_totals = evabs.sel(
                    step="24 hours"
                )
            except KeyError:
                daily_totals = evabs.isel(
                    step=-1
                )
        else:
            daily_totals = evabs

        # Sum daily evaporation across the available period
        if "time" in daily_totals.dims:
            evaporation_total = daily_totals.sum(
                dim="time"
            )
        else:
            evaporation_total = daily_totals

        evaporation_total = evaporation_total.squeeze()

        # Extract the nearest ERA5-Land grid cell
        site_evaporation = evaporation_total.sel(
            latitude=latitude,
            longitude=longitude,
            method="nearest"
        )

        # ERA5-Land evaporation is typically negative.
        # Convert to positive magnitude and from meters to millimeters.
        evaporation_mm = abs(
            float(site_evaporation.values)
        ) * 1000

        record[f"Evaporation_{year}_mm"] = evaporation_mm

        ds.close()

    evaporation_at_locations.append(record)


# Convert results to a DataFrame
df_evaporation_locations = pd.DataFrame(
    evaporation_at_locations
)

df_evaporation_locations = df_evaporation_locations.sort_values(
    by="Evaporation_2024_mm",
    ascending=False
)


# Print results
print("\nCumulative bare-soil evaporation at sampling locations:")

print(
    df_evaporation_locations.to_string(
        index=False,
        float_format=lambda x: f"{x:.2f}"
    )
)
```
